```python
paragraph = """Our university launched a new student study app!!!
The app helps learners organise notes, deadlines, and revision tasks.
Many students say it is AMAZING!!! and saves time.
Visit https://www.studyhelper.example/resources for more details???
Some learners love the simple design and helpful reminders!
However, other students find the app confusing and slow.
A few users reported frustrating errors during registration!!!
The support team fixed several problems and improved the instructions.
Read updates at http://campus.example/news?topic=app&year=2026.
Overall, the feedback is mixed, but students hope the app will improve."""

print(paragraph)
```

```python
import re
```

```python
punctuation_removed = re.sub(r"[^\w\s]", "", paragraph)

print(punctuation_removed)
```

```python
urls = re.findall(r"https?://[^\s]+", paragraph)

print("Found URLs:", urls)
```

```python
#remove URL's
no_urls = re.sub(r"https?://[^\s]+", "", paragraph)

print(no_urls)
```

```python
#remove punctuation, unnecesary space, and tabs
clean_text = re.sub(r"[^\w\s]", "", no_urls)

clean_text = re.sub(r"[ \t]+", " ", clean_text)

clean_text = "\n".join(line.strip() for line in clean_text.splitlines())

print(clean_text)
```

```python
#count URL's and lines
print("Number of URLs:", len(urls))
print("Clean text lines:", len(clean_text.splitlines()))
```

```python
import nltk

nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(no_urls.replace("\n", " "))

print("Sentences:", len(sentences))

for sentence in sentences:
    print("-", sentence)
```

```python
from collections import Counter
```

```python
stop = {"the", "a", "an", "and", "is", "it", "to", "for", "of", "in", "but", "at", "on", "our", "some"}

words = re.findall(r"[a-z]+", no_urls.lower())

frequencies = Counter(w for w in words if w not in stop)

print(frequencies)
```

```python
#gives each sentence a score
scores = []

for sentence in sentences:
    tokens = re.findall(r"[a-z]+", sentence.lower())
    score = sum(frequencies[w] for w in tokens if w not in stop)
    scores.append(score / max(1, len(tokens)))

print(scores)
```

```python
chosen = sorted(range(len(sentences)), key=lambda i: scores[i], reverse=True)[:3]

extractive_summary = " ".join(sentences[i] for i in sorted(chosen))

print("EXTRACTIVE SUMMARY:\n", extractive_summary)
```

```python
#change [:3] to [:2]
chosen = sorted(range(len(sentences)), key=lambda i: scores[i], reverse=True)[:2]

extractive_summary = " ".join(sentences[i] for i in sorted(chosen))

print("EXTRACTIVE SUMMARY:\n", extractive_summary)
```

```python
import nltk

nltk.download("vader_lexicon")
```

```python
from nltk.sentiment import SentimentIntensityAnalyzer

analyser = SentimentIntensityAnalyzer()
```

```python
def sentiment_label(text):
    score = analyser.polarity_scores(text)["compound"]

    if score >= 0.05:
        return "Positive", score
    elif score <= -0.05:
        return "Negative", score
    else:
        return "Neutral", score
```

0.05 to +1 - Positive
-0.05 to 0.05 - Neutral
-1 to -0.05	- Negative

```python
import pandas as pd

rows = []

for sentence in sentences:
    label, score = sentiment_label(sentence)

    rows.append({
        "Sentence": sentence,
        "Sentiment": label,
        "Score": round(score, 3)
    })

results = pd.DataFrame(rows)

display(results)
```
